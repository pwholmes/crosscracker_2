```python
import chromadb
from chromadb import Collection
from chromadb.utils.embedding_functions import ONNXMiniLM_L6_V2

CHROMA_HOST = 'localhost'
CHROMA_PORT = 8001
COLLECTION_NAME = "crossword_1"
EXECUTION_PROVIDER = 'CUDAExecutionProvider'

# Connect to the server
client = chromadb.HttpClient(host=CHROMA_HOST, port=CHROMA_PORT)

# Create the embedding function
gpu_ef = ONNXMiniLM_L6_V2(preferred_providers=[EXECUTION_PROVIDER])

# Get the collection
collection: Collection = client.get_or_create_collection(name=COLLECTION_NAME, embedding_function=gpu_ef)  #type: ignore
```

```python
query_text = "little dogs"

# Run a query
results = collection.query(
    query_texts=[query_text],
    n_results=7
)

# Show the results
#results
if results["documents"] and results["distances"] and results["metadatas"]:
    for distance,clue,answer in zip(results["distances"][0], results["documents"][0], results["metadatas"][0]):
        print(f"{distance:.2f} {answer["answer"]} : {clue}")
```
